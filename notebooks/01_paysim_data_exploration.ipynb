### Step 43: Fraud Transaction Balance Analysis

- **Purpose:** Select fraud-labeled transactions and calculate the difference between the sender's balance before and after each transaction.
- **Result:** The first five rows contain TRANSFER and CASH_OUT transactions. For example, in a transaction with an amount of 181.0, the sender's balance decreased from 181.0 to 0.0, resulting in a balance change of 181.0.
- **Company Use:** Banking fraud investigation teams can compare changes in sender balances with transaction amounts to investigate suspicious activities and identify potential balance inconsistencies.

```python
fraud_balance_analysis = df.loc[
    df["isFraud"] == 1,
    ["type", "amount", "oldbalanceOrg", "newbalanceOrig"]
].copy()

fraud_balance_analysis["balance_change"] = (
    fraud_balance_analysis["oldbalanceOrg"]
    - fraud_balance_analysis["newbalanceOrig"]
)
fraud_balance_analysis.head()
```

### Step 44: Fraud Transaction Amount vs Balance Change

- **Purpose:** Compare transaction amounts with the decrease in the sender's balance for fraud-labeled transactions.
- **Result:** In the first five displayed transactions, the `amount_vs_balance` value is 0.0, indicating that the transaction amount matches the sender's balance change.
- **Company Use:** Banking analytics teams can use this consistency check to investigate differences between transaction amounts and balance movements. However, this result alone does not confirm fraud.

```python
fraud_balance_analysis["amount_vs_balance"] = (
    fraud_balance_analysis["amount"]
    - fraud_balance_analysis["balance_change"]
)
fraud_balance_analysis.head()
```

### Step 45: Fraud Transaction Balance Mismatch Analysis

- **Purpose:** Compare transaction amounts with sender balance changes across all fraud-labeled transactions to identify transactions with mismatches.
- **Result:** Out of 8,213 fraud-labeled transactions, 69 transactions had balance mismatches, while 8,144 transactions had matching balances. The mismatch percentage was approximately 0.84%.
- **Company Use:** Banking teams can identify transactions with balance mismatches for reconciliation, auditing, and further investigation. A mismatch is an indicator for investigation; it does not, by itself, prove fraud.

```python
balance_mismatch = fraud_balance_analysis[
    fraud_balance_analysis["amount_vs_balance"] != 0
]
print("Toatl fraud-labeled transactions:", len(fraud_balance_analysis))
print("Transaction with balance mismatch:", len(balance_mismatch))
print("Transactions with matching balance:", len(fraud_balance_analysis) - len(balance_mismatch))
```

### Step 46: Fraud Transactions by Transaction Type

- **Purpose:** Calculate the count and percentage of fraud-labeled transactions for each transaction type.
- **Result:** Out of 8,213 fraud-labeled transactions, 4,116 were CASH_OUT transactions (50.12%), and 4,097 were TRANSFER transactions (49.88%). Both transaction types account for nearly equal proportions of fraud-labeled transactions.
- **Company Use:** Banking analytics teams can understand the distribution of fraud-labeled transactions by transaction type and prepare reports to support fraud monitoring and investigation.

```python
fraud_type_summary = (
    fraud_balance_analysis["type"]
    .value_counts()
    .to_frame(name="fraud_count")
)
fraud_type_summary["percentage"] = (
    fraud_type_summary["fraud_count"]
    / len(fraud_balance_analysis)
) * 100
fraud_type_summary["percentage"] = fraud_type_summary["percentage"].round(2)
fraud_type_summary
```

### Step 47: Fraud Amount by Transaction Type

- **Purpose:** Calculate the total amount of fraud-labeled transactions for each transaction type and its percentage of the overall fraud-labeled transaction amount.
- **Result:** CASH_OUT transactions account for approximately 5.989 billion (49.68%), while TRANSFER transactions account for approximately 6.067 billion (50.32%). TRANSFER contributes a slightly higher share of the total fraud-labeled transaction amount.
- **Company Use:** Banking analytics teams can analyze the distribution of fraud-labeled transaction amounts to support investigations and reporting. These amounts should not be assumed to represent actual financial losses.

```python
fraud_amount_summary = (
    fraud_balance_analysis
    .groupby("type")["amount"]
    .sum()
    .to_frame(name="total_fraud_amount")
)
fraud_amount_summary["percentage"] = (
    fraud_amount_summary["total_fraud_amount"]
    / fraud_amount_summary["total_fraud_amount"].sum()
) * 100
fraud_amount_summary["percentage"] = (
    fraud_amount_summary["percentage"].round(2)
)
fraud_amount_summary
```

### Step 48: Fraud Transaction Sender Balance Analysis

- **Purpose:** Calculate the transaction count, average old balance, and average new balance for fraud-labeled transactions by transaction type.
- **Result:** CASH_OUT includes 4,116 transactions, with an average old balance of 1,453,869.05 and an average new balance of 72.59. TRANSFER includes 4,097 transactions, with an average old balance of 1,846,374.19 and an average new balance of 385,604.57.
- **Company Use:** Banking analytics teams can compare sender balance patterns across transaction types to identify cases for further investigation. Average values alone do not prove fraud.

```python
fraud_sender_balance = (
    fraud_balance_analysis
    .groupby("type")
    .agg(
        transaction_count=("amount", "count"),
        average_old_balance=("oldbalanceOrg", "mean"),
        average_new_balance=("newbalanceOrig", "mean")
    )
    .round(2)
)
fraud_sender_balance
```

### Step 49: Fraud Transaction Sender Zero Balance Analysis

- **Purpose:** Calculate the count and percentage of fraud-labeled transactions in which the sender's balance is zero after the transaction.
- **Result:** Out of 8,213 fraud-labeled transactions, 8,053 (98.05%) had a zero sender balance after the transaction, while 160 (1.95%) had a non-zero balance.
- **Company Use:** Banking analytics teams can compare the pattern of zero sender balances with other risk indicators to investigate suspicious transactions. A zero balance is one signal; it does not, by itself, prove fraud.

```python
zero_balance_count = (
    fraud_balance_analysis["newbalanceOrig"] == 0
).sum()
nonzero_balance_count = (
    fraud_balance_analysis["newbalanceOrig"] != 0
).sum()
total_fraud = len(fraud_balance_analysis)
print(f"Total fraud-labeled transaction: {total_fraud:,}(100.00%)")
print(
    f"Zero sender balance: {zero_balance_count:,}"
    f"({zero_balance_count / total_fraud * 100:.2f}%)"
)
print(
    f"Non-zero sender balance:{nonzero_balance_count:,}"
    f"({nonzero_balance_count / total_fraud * 100:.2f}%)"
)
```

### Step 50: Fraud Transactions — Zero Balance by Transaction Type

- **Purpose:** Calculate the count and percentage of fraud-labeled transactions with a zero sender balance after the transaction, grouped by transaction type.
- **Result:** Among CASH_OUT transactions, 4,115 out of 4,116 (99.98%) had a zero sender balance. Among TRANSFER transactions, 3,938 out of 4,097 (96.12%) had a zero sender balance.
- **Company Use:** Banking analytics teams can compare zero-balance patterns across transaction types to identify cases for further investigation. A zero balance is one indicator; it does not, by itself, prove fraud.

```python
zero_balance_summary = (
    fraud_balance_analysis
    .assign(
        zero_balance=fraud_balance_analysis["newbalanceOrig"] == 0
    )
    .groupby("type")
    .agg(
        total_transactions=("zero_balance", "size"),
        zero_balance_count=("zero_balance", "sum")
    )
)
zero_balance_summary["zero_balance_percentage"] = (
    zero_balance_summary["zero_balance_count"]
    / zero_balance_summary["total_transactions"]
    * 100
).round(2)
zero_balance_summary
```

### Step 51: Fraud Transaction Amount by Balance Status

- **Purpose:** Calculate the transaction count, average amount, and total amount for fraud-labeled transactions grouped by whether the sender's balance is zero or non-zero.
- **Result:** The Non-Zero Balance group contains 160 transactions, with an average amount of 9,367,638.00 and a total amount of approximately 1.499 billion. The Zero Balance group contains 8,053 transactions, with an average amount of 1,311,013.70 and a total amount of approximately 10.558 billion.
- **Company Use:** Banking analytics teams can compare transaction amounts by balance status to identify high-value transactions for further investigation. This pattern alone does not prove fraud.

```python
fraud_balance_analysis["balance_status"] = (
    fraud_balance_analysis["newbalanceOrig"]
    .eq(0)
    .map({
        True: "Zero Balance",
        False: "Non-Zero Balance"
    })
)

fraud_balance_amount_summary = (
    fraud_balance_analysis
    .groupby("balance_status")
    .agg(
        transaction_count=("amount", "count"),
        average_amount=("amount", "mean"),
        total_amount=("amount", "sum")
    )
    .round(2)
)

fraud_balance_amount_summary
```
